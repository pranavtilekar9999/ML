## Experiment 04
**Aim:** Given the following data, which specifies classifications for nine combinations of VAR1 and VAR2, predict a classification for a case where VAR1 = 0.906 and VAR2 = 0.606, using the result of K-Means clustering with 3 means (3 centroids).

```python
import numpy as np
from sklearn.cluster import KMeans

X = np.array([
    [1.713, 1.586], [0.180, 1.786], [0.353, 1.240],
    [0.940, 1.566], [1.486, 0.759], [1.266, 1.106],
    [1.540, 0.419], [0.459, 1.799], [0.773, 0.186]
])

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X)

print("Cluster Labels:", kmeans.labels_ + 1)
print("Centroids:")
print(kmeans.cluster_centers_)

new_data = np.array([[0.906, 0.606]])
cluster = kmeans.predict(new_data)
print("Predicted Cluster:", cluster[0] + 1)
```
