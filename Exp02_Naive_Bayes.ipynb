## Experiment 02
**Aim:** Implement Naive Bayes theorem to classify the English text

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

texts = [
    "I love this movie", "This movie is excellent",
    "The movie was amazing", "I enjoyed this film",
    "This is a wonderful movie", "The acting was great",
    "I liked this movie", "This film was fantastic",
    "The story was interesting", "It was an excellent experience",
    "I hate this movie", "This movie is terrible",
    "The movie was boring", "I disliked this film",
    "This is a horrible movie", "The acting was bad",
    "I did not like this movie", "This film was disappointing",
    "The story was boring", "It was a terrible experience"
]
labels = [1]*10 + [0]*10

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

model = MultinomialNB()
model.fit(X, labels)

test_texts = [
    "I love this excellent film",
    "The movie was boring",
    "This is a fantastic movie",
    "I disliked this terrible movie",
    "The acting was great"
]
X_test = vectorizer.transform(test_texts)
predictions = model.predict(X_test)
probabilities = model.predict_proba(X_test)

for i in range(len(test_texts)):
    result = "Positive" if predictions[i] == 1 else "Negative"
    print("\nText:", test_texts[i])
    print("Prediction:", result)
    print("Positive:", round(probabilities[i][1], 3))
    print("Negative:", round(probabilities[i][0], 3))
```
